````python
# creating a list of integers between 0 and 100
# first create an object to assign the integers to

integers = list(range(0,100))
# this creates a list of integers. the range function makes
# the numbers between 0 and 100 the items in the list

# then i have to create another object to "hold" the
# even numbers that I actually want it to count:

evens = 0

# the =0 part effectively sets "evens" to zero, that way
# it has a place to start adding

# then i create a loop for the computer to start checking
# all the numbers in the range:
for numbers in integers:
# this creates the for loop ^
  if numbers % 2 == 0:
# this tells the computer to check all the numbers in the
# list for if they are divisible by 2 with a remainder
# of zero (ie. an even number)
    evens += numbers
# this tells the computer to add all the even numbers
# it finds to the "evens" object
print (evens)

# this final "print" command needs to be directly lined
# up with the for loop command for the code to work

````

````python
# Create a list of all integers from 100 to 200 (inclusive)
all_integers = list(range(100, 201))

# Initialize a variable to store the sum of odd integers
sum_of_odd_integers = 0

# Iterate through the list and sum the odd integers
for number in all_integers:
    if number % 2 != 0: # Check if the number is odd
        sum_of_odd_integers += number

# Print the final sum
print(f"The sum of all odd integers between 100 and 200 is: {sum_of_odd_integers}")
````
